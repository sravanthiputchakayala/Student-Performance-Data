# Univariate Analysis (Distributions)

```python
# correlation matrix
corr_matrix = df.corr()
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
#Histograms
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(df['studytime'], kde=True, color='blue')  # kde=True adds a smoothed curve
plt.title('Histogram of Numerical Column')
plt.xlabel('Numerical Column')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(20, 20))
df.hist()
plt.show()
```

```python
#Box Plots
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['studytime'], color='green')
plt.title('Box Plot of Numerical Column')
plt.xlabel('Numerical Column')
plt.show()
```

```python
plt.figure(figsize=(15, 8))
df.boxplot()
plt.show()
```

```python
#Density Plot
plt.figure(figsize=(10, 6))
sns.kdeplot(df['studytime'], shade=True, color='purple')
plt.title('Density Plot of Column ')
plt.xlabel('Age')
plt.ylabel('Density')
plt.show()
```

```python
plt.figure(figsize=(29, 10))
df.plot.kde()
plt.show()
```

```python
#pie charts
category_counts = df['studytime'].value_counts()
category_counts.plot.pie(autopct='%1.1f%%', figsize=(10, 10))
plt.title('Pie Chart of Category Column Studytime')
plt.ylabel('')
plt.show()
```

```python
print(df['studytime'].skew()) #right #skewness
```

```python
print(df['studytime'].kurt()) #kurtosis
```

```python
#outliners
Q1 = df['studytime'].quantile(0.25)
Q3 = df['studytime'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[(df['age'] < lower_bound) | (df['age'] > upper_bound)]
print(outliers)
```

```python
sns.countplot(x='studytime', data=df)
```

```python
age_counts = df['studytime'].value_counts().reset_index()
age_counts.columns = ['studytime', 'count']  # Rename columns for clarity
plt.figure(figsize=(6, 6))
sns.barplot(x='studytime', y='count', data=age_counts, palette='viridis')

```
